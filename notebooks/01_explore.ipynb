```python
import pandas as pd

df = pd.read_csv("../data/raw/ptb-xl/ptbxl_database.csv", index_col="ecg_id")

print("Shape (ECGs, columns):", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nUnique patients:", df["patient_id"].nunique())

print("\n--- Age ---")
print(df["age"].describe())
print("Age 300 (anonymised >89):", (df["age"] >= 300).sum())

print("\n--- Pacemaker ---")
print(df["pacemaker"].value_counts(dropna=False))

print("\n--- Noise flags (number of ECGs) ---")
for col in ["baseline_drift", "static_noise", "burst_noise", "electrodes_problems"]:
    print(f"{col}: {df[col].notna().sum()}")
print("Any noise:", df[["baseline_drift", "static_noise", "burst_noise", "electrodes_problems"]].notna().any(axis=1).sum())

print("\n--- Example diagnosis codes ---")
print(df["scp_codes"].head())

print("\n--- ECGs per fold ---")
print(df["strat_fold"].value_counts().sort_index())
```
